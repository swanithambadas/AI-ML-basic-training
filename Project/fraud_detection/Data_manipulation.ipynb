### Data preparation for model training 

```python
# converting the datetime format into actual date and and actual hours format
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'])
df['hour'] = df['trans_date_trans_time'].dt.hour
df['day_of_week'] = df['trans_date_trans_time'].dt.dayofweek
```

```python
# 2. Calculate User Age from DOB
# XGBoost can't read a '1985-10-25' string. It needs a number (e.g., 38).
df['dob'] = pd.to_datetime(df['dob'])
# Using 2026 as the current year based on your system context
df['age'] = 2026 - df['dob'].dt.year 
```

```python
df.head()
```

```python
print("Calculating User-to-Merchant distances...")
def haversine_vectorized(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2.0)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2.0)**2
    return 6371 * 2 * np.arcsin(np.sqrt(a))
    
df['dist_to_merchant_km'] = haversine_vectorized(
        df['lat'], df['long'], df['merch_lat'], df['merch_long']
    )
```

```python
# Target Encode 'category'
cat_target_means = df.groupby('category')['is_fraud'].mean()
df['category_encoded'] = df['category'].map(cat_target_means)
    
    # Target Encode 'gender'
gender_target_means = df.groupby('gender')['is_fraud'].mean()
df['gender_encoded'] = df['gender'].map(gender_target_means)
```

```python
avg_dist_legit = df[df['is_fraud'] == 0]['dist_to_merchant_km'].mean()
avg_dist_fraud = df[df['is_fraud'] == 1]['dist_to_merchant_km'].mean()
print(f"Distance Signal: Legitimate avg = {avg_dist_legit:.2f}km | Fraud avg = {avg_dist_fraud:.2f}km")
```

```python
# Insight B: Highest Risk Categories (Percentage of transactions that are fraud)
fraud_rate_by_cat = df.groupby('category')['is_fraud'].mean().sort_values(ascending=False) * 100
print(f"\nTop 3 Riskiest Categories (Highest % of fraud):")
print(fraud_rate_by_cat.head(3).to_string())
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
correlations = df[numeric_cols].corr()['is_fraud'].sort_values(ascending=False)
print("\nFeature Correlation to Fraud (Closer to 1 or -1 is a stronger signal):")
print(correlations.drop('is_fraud').to_string())
```

```python
fraud_counts = df['is_fraud'].value_counts().reset_index()
fraud_counts.columns = ['is_fraud', 'count']
fraud_counts['is_fraud'] = fraud_counts['is_fraud'].map({0: 'Legitimate', 1: 'Fraud'})
    
fig1 = px.pie(fraud_counts, names='is_fraud', values='count', 
                  title="1. Class Imbalance (Legitimate vs Fraud)",
                  color='is_fraud', color_discrete_map={'Legitimate':'blue', 'Fraud':'red'})
fig1.show()
```

```python
fig2 = px.box(df, x='is_fraud', y='amt', color='is_fraud',
                  title="2. Transaction Amount Distribution (Log Scale)",
                  labels={'is_fraud': 'Is Fraud?', 'amt': 'Transaction Amount ($)'},
                  log_y=True, color_discrete_map={0:'blue', 1:'red'})
    # Update x-axis labels for clarity
fig2.update_xaxes(tickvals=[0, 1], ticktext=['Legitimate', 'Fraud'])
fig2.show()
```

```python
fraud_only = df[df['is_fraud'] == 1]
category_counts = fraud_only['category'].value_counts().reset_index()
category_counts.columns = ['category', 'fraud_count']
    
fig3 = px.bar(category_counts, x='category', y='fraud_count',
                  title="3. Fraudulent Transactions by Merchant Category",
                  labels={'category': 'Merchant Category', 'fraud_count': 'Number of Fraud Events'},
                  color='fraud_count', color_continuous_scale='Reds')
fig3.update_layout(xaxis_tickangle=-45)
fig3.show()
```

```python
hourly_fraud = fraud_only.groupby('hour').size().reset_index(name='fraud_count')
fig4 = px.line(hourly_fraud, x='hour', y='fraud_count', markers=True,
                   title="4. Fraudulent Transactions by Time of Day (Hour)",
                   labels={'hour': 'Hour of Day (0-23)', 'fraud_count': 'Number of Fraud Events'},
                   line_shape='spline')
fig4.show()
```
