```python
frameworks = ["Spring Boot", "LangGraph", "FastAPI"]

# --- Forward while loop ---
print("--- Forward while loop ---")
i = 0
while i < len(frameworks):
    print(f"Index {i}: {frameworks[i]}")
    i += 1  # Increment to move forward


# --- Reverse while loop ---
print("\n--- Reverse while loop ---")
i = len(frameworks) - 1  # Start at last index
while i >= 0:
    print(f"Index {i}: {frameworks[i]}")
    i -= 1  # Decrement to move backward
```

```python
# Demonstrating return vs print
def check_prime_range(num):
    # range(2, num) doesn't include 1 and the number itself
    for n in range(2, num):
        if num % n == 0:
            return False  # Return allows us to reuse this boolean value
    return True

# Storing the returned value for reuse
is_seven_prime = check_prime_range(7)
print(f"Is 7 prime? {is_seven_prime}")

# Print returns None
def print_only(num):
    print(f"The number is {num}")

result = print_only(5)
print(f"The variable 'result' holds: {result}")
```

```python
# Generator function to yield cubes
def gen_cubes(n):
    for num in range(n):
        yield num ** 3  # Pauses and yields value on the fly

# Using a for loop to decide the actual parameter and iterate
print("Cubes up to 4:")
for x in gen_cubes(4):
    print(x)

# Using next() on a generator
print("\nUsing next():")
cube_gen = gen_cubes(3)
print(next(cube_gen)) # 0
print(next(cube_gen)) # 1
print(next(cube_gen)) # 8
```

```python
# Using map with a lambda function to convert temperatures
celsius_temps = [0, 10, 20, 30]

# map(function, list of values)
fahrenheit_temps = list(map(lambda c: (c * 9/5) + 32, celsius_temps))
print("Mapped Temperatures:", fahrenheit_temps)

# Using filter to only keep temperatures above 15 degrees Celsius
warm_temps = list(filter(lambda c: c > 15, celsius_temps))
print("Filtered Warm Temperatures:", warm_temps)
```

```python
from dataclasses import dataclass

# ==========================================
# 1. ENCAPSULATION & COMPOSITION
# ==========================================

class Engine:
    def start(self):
        print("Engine starting")

class Car:
    def __init__(self):
        # The Car class is composed of an Engine object.
        self.engine = Engine()
        
    def start(self):
        print("start engine first")
        # Calling the component's method
        self.engine.start()
        print("Car has started")

# ==========================================
# 2. DYNAMIC EXTENSION (WRAPPER)
# ==========================================

class Text:
    def render(self):
        return "Hello"

class HeadingWrapper:
    def __init__(self, wrapped_text):
        self.wrappend = wrapped_text
        
    def render(self):
        # Dynamically extending the functionality of the wrapped object
        return f"<h>{self.wrappend.render()}</h>"

# ==========================================
# 3. POLYMORPHISM & DUCK TYPING
# ==========================================

def display_render(item):
    # Dynamic typing adapts based on the input provided.
    # As long as the object has a 'render' method, it works (Duck Typing).
    print(item.render())

# ==========================================
# 4. CLASS vs INSTANCE VARIABLES & METHODS
# ==========================================

class Student:
    # Class Variable: Assigned to the whole class, defines scope for a particular variable.
    school = "ABC School"
    _instance_count = 0  # Used for tracking count
    
    def __init__(self, name):
        # Instance Variable: Object-specific variable.
        self.name = name
        Student._instance_count += 1
        
    # Instance Method
    def get_details(self):
        # Takes 'self' as the first argument.
        # Can access both instance variables (self.name) and class variables (self.school).
        return f"Student: {self.name}, School: {self.school}"
        
    # Class Method
    @classmethod
    def get_instance_count(cls):
        # Takes 'cls' as the input parameter.
        # Can modify the class as well.
        # Can access only class variables; cannot access instance variables.
        return cls._instance_count

# ==========================================
# 5. DECORATORS & DATACLASSES
# ==========================================

# @dataclass is a decorator that provides key-value pairs instead of manual 'self' assignments and allows for strict data type checking.
@dataclass
class Person:
    name: str
    age: int
    
    # A byproduct of decorators like @dataclass
    def __post_init__(self):
        # This runs immediately after the generated __init__ completes.
        self.name = self.name.strip().title()

# ==========================================
# TESTING THE CONCEPTS
# ==========================================

if __name__ == "__main__":
    print("--- Composition ---")
    my_car = Car()
    my_car.start()
    
    print("\n--- Dynamic Extension & Duck Typing ---")
    basic_text = Text()
    heading_text = HeadingWrapper(basic_text)
    display_render(basic_text)    # Outputs: Hello
    display_render(heading_text)  # Outputs: <h>Hello</h>
    
    print("\n--- Variables & Methods ---")
    s1 = Student("Alice")
    s2 = Student("Bob")
    
    print(s1.get_details())
    
    # Modifying class variable via the class
    Student.school = "XYZ School"
    print(f"After Student.school change -> {s1.get_details()}")
    
    # Attempting to modify class variable via instance (creates an instance variable instead)
    s1.school = "LMN School"
    print(f"s1 school: {s1.school}")
    print(f"s2 school (unchanged): {s2.school}")
    
    # Accessing Class Method
    print(f"Total students created: {Student.get_instance_count()}")
    
    print("\n--- Dataclasses ---")
    p1 = Person(name="  john doe  ", age=30)
    # The __post_init__ cleaned up the name string
    print(f"Person created: {p1.name}, Age: {p1.age}")
```

```python
# ==========================================
# STATIC METHODS
# ==========================================

class Student:
    school = "ABC School"
    
    def __init__(self, name):
        self.name = name

    # @staticmethod is a decorator that modifies the function to belong to the class
    # without requiring an instance reference.
    @staticmethod
    def is_school_day(day_of_week):
        """
        Notice that this method does NOT take 'self' or 'cls' as an argument.
        It behaves just like a regular Python function, but it is housed inside 
        the class because it logically relates to Students.
        """
        weekend = ["saturday", "sunday"]
        if day_of_week.lower() in weekend:
            return False
        return True

# ==========================================
# TESTING THE CONCEPT
# ==========================================

if __name__ == "__main__":
    # You can call a static method directly on the Class itself
    print(f"Is Monday a school day? {Student.is_school_day('monday')}")
    
    # You can also call it on an instance, though calling it on the class is preferred
    s1 = Student("Alice")
    print(f"Is Saturday a school day? {s1.is_school_day('saturday')}")
```

##### Encapsulation & Composition: Composition is demonstrated by a Car class that initializes an Engine class within its __init__ method. The Car class's start method prints "start engine first", calls self.engine.start(), and then prints "Car has started". The Engine class has a start method that prints "Engine starting".  
##### Dynamic Extension: Functionality can be dynamically extended by wrapping objects. A Text class with a render method returning "Hello" can be wrapped by another class that modifies the return to f"<h>{self.wrappend.render()}</h>".  Polymorphism & Duck Typing: Dynamic typing adapts dynamically based on the input provided.  
##### Variables:Class Variables: These define the scope of a particular variable and are assigned to the whole class. For example, school = "ABC School" inside the Student class is a class variable.  Instance Variables: These are object-specific variables, such as self.name = name within the __init__ method.  
##### Methods:Instance Methods: These methods take self as the first argument and can access both instance and class variables.  Class Methods: Annotated with @classmethod, these take cls as an input parameter. They can modify the class, can only access class variables, and cannot access instance variables. An example is returning cls values like an instance count.  
##### Static Methods & Decorators: Decorators modify and enhance existing functions' functionality to behave in a certain way without changing the original method.  
##### Dataclasses: The @dataclass decorator uses key-value pairs instead of self and allows for strict data type checking. Pydantic utilizes dataclasses. A byproduct of this decorator is the ability to use def __post_init__(self):.  
##### Single Responsibility Principle (SRP): When considering how long a class is, it is important to follow SRP, meaning there should be only one responsibility per class.
