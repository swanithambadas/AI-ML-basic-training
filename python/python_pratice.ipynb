```python
# --- Forward Range ---
print("--- Forward Range (0 to 4) ---")
for i in range(5):
    print(f"Index: {i}")

print("\n--- Forward Range with Custom Start/Stop (1 to 5) ---")
for i in range(1, 6):
    print(f"Step {i}")


# --- Reverse Range ---
print("\n--- Reverse Range (5 down to 1) ---")
# start at 5, stop before 0 (which is 1), step by -1
for i in range(5, 0, -1):
    print(f"Countdown: {i}")

print("\n--- Reverse Range with Skipping (10 down to 2 by 2s) ---")
for i in range(10, 0, -2):
    print(f"Even Countdown: {i}")
```

```python
frameworks = ["Spring Boot", "LangGraph", "FastAPI"]

# --- Forward while loop ---
print("--- Forward while loop ---")
i = 0
while i < len(frameworks):
    print(f"Index {i}: {frameworks[i]}")
    i += 1  # Increment to move forward


# --- Reverse while loop ---
print("\n--- Reverse while loop ---")
i = len(frameworks) - 1  # Start at last index
while i >= 0:
    print(f"Index {i}: {frameworks[i]}")
    i -= 1  # Decrement to move backward
```

```python
tech_stack = ["Python", "Java", "SQL", "React"]

# --- Forward List Iteration with Index tracking ---
print("--- Forward List with enumerate() ---")
for index, item in enumerate(tech_stack):
    print(f"Item {index}: {item}")


# --- Reverse List Iteration (Option A: Slicing [::-1]) ---
print("\n--- Reverse List with Slicing ---")
for item in tech_stack[::-1]:
    print(f"Reversed Item: {item}")


# --- Reverse List Iteration (Option B: range index) ---
print("\n--- Reverse List with Reverse range() ---")
for i in range(len(tech_stack) - 1, -1, -1):
    print(f"Index {i}: {tech_stack[i]}")
```

```python
# Demonstrating return vs print
def check_prime_range(num):
    # range(2, num) doesn't include 1 and the number itself
    for n in range(2, num):
        if num % n == 0:
            return False  # Return allows us to reuse this boolean value
    return True

# Storing the returned value for reuse
is_seven_prime = check_prime_range(7)
print(f"Is 7 prime? {is_seven_prime}")

# Print returns None
def print_only(num):
    print(f"The number is {num}")

result = print_only(5)
print(f"The variable 'result' holds: {result}")
```

```python
# Generator function to yield cubes
def gen_cubes(n):
    for num in range(n):
        yield num ** 3  # Pauses and yields value on the fly

# Using a for loop to decide the actual parameter and iterate
print("Cubes up to 4:")
for x in gen_cubes(4):
    print(x)

# Using next() on a generator
print("\nUsing next():")
cube_gen = gen_cubes(3)
print(next(cube_gen)) # 0
print(next(cube_gen)) # 1
print(next(cube_gen)) # 8
```

```python
# Using map with a lambda function to convert temperatures
celsius_temps = [0, 10, 20, 30]

# map(function, list of values)
fahrenheit_temps = list(map(lambda c: (c * 9/5) + 32, celsius_temps))
print("Mapped Temperatures:", fahrenheit_temps)

# Using filter to only keep temperatures above 15 degrees Celsius
warm_temps = list(filter(lambda c: c > 15, celsius_temps))
print("Filtered Warm Temperatures:", warm_temps)
```

```python
# Standard function approach
def add_numbers(x, y):
    return x + y

print("Standard def:", add_numbers(5, 3))

# Equivalent Lambda function approach
# It takes x and y as arguments, and returns x + y
add_lambda = lambda x, y: x + y

print("Lambda:", add_lambda(5, 3))

# Another example: A lambda to square a number
square = lambda x: x ** 2
print("Squared:", square(4))
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Using reduce with a lambda to sum all numbers in the list
# reduce(lambda a, b: a + b, list)
total_sum = reduce(lambda a, b: a + b, numbers)

print("Reduced Single Value (Sum):", total_sum)
```

```python
# 1. Writing to a file (Creates the file or overwrites it)
with open("iliad_notes.txt", "w") as f:
    f.write("The Trojan War was a major conflict.\n")
    f.write("Achilles was a legendary warrior.\n")

# 2. Appending to a file (Preserves previous content)
with open("iliad_notes.txt", "a") as f:
    f.write("Hector was the prince of Troy.\n")

# 3. Reading the file completely
with open("iliad_notes.txt", "r") as f:
    content = f.read()
    print("--- Full Content ---")
    print(content)

# 4. Using readline, readlines, and read(10)
with open("iliad_notes.txt", "r") as f:
    print("--- Reading Operations ---")
    # Gives the first 10 characters
    print("First 10 chars:", f.read(10)) 
    
    # Reads the rest of the current line and removes extra spaces
    print("Next line stripped:", f.readline().strip()) 
    
    # Reads all remaining lines into a list format
    print("Remaining lines as list:", f.readlines()) 
    
# 5. Iterating line by line to clean extra spaces
with open("iliad_notes.txt", "r") as f:
    print("\n--- Iterating and Stripping ---")
    for line in f:
        print("line:", line.strip())
```

```python
def read_log_file(filename):
    try:
        # Code to execute
        f = open(filename, "r")
        content = f.read()
    except FileNotFoundError:
        # Executes if 'try' fails
        print(f"Error: The file {filename} does not exist.")
    else:
        # Executes if 'try' is successful
        print("File read successfully. Length:", len(content))
        f.close()
    finally:
        # Executes in any case
        print("File operation check complete.")

# Testing the block with a non-existent file
read_log_file("missing_odyssey_scroll.txt")
```
