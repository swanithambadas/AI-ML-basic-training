```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:       # 'elif' checks subsequent conditions if 'if' is False
    print("Grade: B")
else:                   # 'else' catches anything that didn't match above
    print("Grade: C")
```

```python
# --- Forward Range ---
print("--- Forward Range (0 to 4) ---")
for i in range(5):
    print(f"Index: {i}")

print("\n--- Forward Range with Custom Start/Stop (1 to 5) ---")
for i in range(1, 6):
    print(f"Step {i}")


# --- Reverse Range ---
print("\n--- Reverse Range (5 down to 1) ---")
# start at 5, stop before 0 (which is 1), step by -1
for i in range(5, 0, -1):
    print(f"Countdown: {i}")

print("\n--- Reverse Range with Skipping (10 down to 2 by 2s) ---")
for i in range(10, 0, -2):
    print(f"Even Countdown: {i}")
```

```python
frameworks = ["Spring Boot", "LangGraph", "FastAPI"]

# --- Forward while loop ---
print("--- Forward while loop ---")
i = 0
while i < len(frameworks):
    print(f"Index {i}: {frameworks[i]}")
    i += 1  # Increment to move forward


# --- Reverse while loop ---
print("\n--- Reverse while loop ---")
i = len(frameworks) - 1  # Start at last index
while i >= 0:
    print(f"Index {i}: {frameworks[i]}")
    i -= 1  # Decrement to move backward
```

```python
tech_stack = ["Python", "Java", "SQL", "React"]

# --- Forward List Iteration with Index tracking ---
print("--- Forward List with enumerate() ---")
for index, item in enumerate(tech_stack):
    print(f"Item {index}: {item}")


# --- Reverse List Iteration (Option A: Slicing [::-1]) ---
print("\n--- Reverse List with Slicing ---")
for item in tech_stack[::-1]:
    print(f"Reversed Item: {item}")


# --- Reverse List Iteration (Option B: range index) ---
print("\n--- Reverse List with Reverse range() ---")
for i in range(len(tech_stack) - 1, -1, -1):
    print(f"Index {i}: {tech_stack[i]}")
```
